```python
import subprocess

print("Inicio ejecución Entrega 1")

notebooks = [
    "Parte-A/Parte_A.ipynb",
    "Parte-B/Entrega_1_Parte_B_completo.ipynb",
    "Parte-C/PARTE_C.ipynb"
]

for nb in notebooks:
    print(f"\nEjecutando {nb}...")

    subprocess.run([
        "jupyter",
        "nbconvert",
        "--to", "notebook",
        "--execute",
        "--inplace",
        "--ExecutePreprocessor.timeout=-1",
        nb
    ], check=True)

print("\nEntrega 1 ejecutada correctamente")
```
